```python
import pandas as pd

file_path = r"/kaggle/input/datasets/bravehart101/sample-supermarket-dataset/SampleSuperstore.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python

# STEP 6: DATA CLEANING AND BUSINESS KPIs

# Keep original data safe
df_clean = df.copy()

# Remove exact duplicate rows
duplicate_count = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates().copy()

# Check for impossible or unusual numeric values
print("DATA CLEANING SUMMARY")
print("-" * 45)
print("Original rows:", len(df))
print("Exact duplicate rows removed:", duplicate_count)
print("Rows after cleaning:", len(df_clean))
print("Missing values remaining:", df_clean.isnull().sum().sum())

print("\nBUSINESS KPIs")
print("-" * 45)

total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders_rows = len(df_clean)
total_quantity = df_clean["Quantity"].sum()
profit_margin = (total_profit / total_sales * 100) if total_sales else 0
average_sales_per_row = df_clean["Sales"].mean()
average_profit_per_row = df_clean["Profit"].mean()

kpis = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity Sold",
        "Number of Transaction Rows",
        "Average Sales per Row",
        "Average Profit per Row",
        "Overall Profit Margin (%)"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        total_orders_rows,
        average_sales_per_row,
        average_profit_per_row,
        profit_margin
    ]
})

display(kpis.round(2))

print("\nNUMERIC DISTRIBUTION")
display(
    df_clean[["Sales", "Quantity", "Discount", "Profit"]]
    .describe()
    .T.round(2)
)
```

```python
import pandas as pd
import numpy as np

# Load the dataset (previous cell mein df load kiya tha)
print("DATASET OVERVIEW")
print("=" * 50)
print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFIRST 5 ROWS")
display(df.head())

print("\nDATA TYPES")
display(df.dtypes.to_frame("Data Type"))

print("\nMISSING VALUES")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

print("\nSTATISTICAL SUMMARY")
display(df.describe(include="all").T)
```

```python

# STEP 7: BUSINESS PERFORMANCE ANALYSIS

print("PERFORMANCE BY CATEGORY")
category_analysis = (
    df_clean.groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values("Sales", ascending=False)
)
display(category_analysis.round(2))

print("\nPERFORMANCE BY REGION")
region_analysis = (
    df_clean.groupby("Region")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_values("Sales", ascending=False)
)
display(region_analysis.round(2))

print("\nPERFORMANCE BY CUSTOMER SEGMENT")
segment_analysis = (
    df_clean.groupby("Segment")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
    .sort_values("Sales", ascending=False)
)
display(segment_analysis.round(2))

print("\nTOP 10 SUB-CATEGORIES BY SALES")
subcategory_analysis = (
    df_clean.groupby("Sub-Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values("Sales", ascending=False)
)
display(subcategory_analysis.head(10).round(2))

print("\nBOTTOM 10 SUB-CATEGORIES BY PROFIT")
display(subcategory_analysis.sort_values("Profit").head(10).round(2))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Chart 1: Sales by category
plt.figure(figsize=(9, 5))
category_analysis["Sales"].sort_values().plot(kind="barh")
plt.title("Total Sales by Product Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

# Chart 2: Profit by category
plt.figure(figsize=(9, 5))
category_analysis["Profit"].sort_values().plot(kind="barh")
plt.title("Total Profit by Product Category")
plt.xlabel("Profit")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

# Chart 3: Sales by region
plt.figure(figsize=(8, 5))
region_analysis["Sales"].sort_values().plot(kind="bar")
plt.title("Sales Comparison Across Regions")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Chart 4: Profit by sub-category
plt.figure(figsize=(10, 7))
subcategory_analysis["Profit"].sort_values().plot(kind="barh")
plt.title("Profitability by Product Sub-Category")
plt.xlabel("Profit")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()

# Chart 5: Sales distribution
plt.figure(figsize=(9, 5))
sns.histplot(df_clean["Sales"], bins=40, kde=True)
plt.title("Distribution of Sales Values")
plt.xlabel("Sales")
plt.tight_layout()
plt.show()

# Chart 6: Discount vs profit
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha=0.35
)
plt.title("Relationship Between Discount and Profit")
plt.tight_layout()
plt.show()
```

```python

numeric_cols = ["Sales", "Quantity", "Discount", "Profit"]
correlation_matrix = df_clean[numeric_cols].corr()

print("CORRELATION MATRIX")
display(correlation_matrix.round(3))

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)
plt.title("Correlation: Sales, Quantity, Discount and Profit")
plt.tight_layout()
plt.show()

print("\nCORRELATION WITH PROFIT")
display(
    correlation_matrix["Profit"]
    .drop("Profit")
    .sort_values()
    .to_frame("Correlation with Profit")
    .round(3)
)
```
